```python
from langchain.agents import create_agent
from langchain.tools import tool
from dotenv import load_dotenv

# 1.加载环境变量
load_dotenv() # 读取.env文件，加入系统环境变量中

# 2.定义工具，就是定义函数，给函数加装饰器 @tool
@tool
def get_weather(x: str) -> str:
    """获取天气"""
    return f"Current weather in {x} is cloudy"



@tool
def square_root(x: float) -> float:
    """求x的平方根"""
    return x ** 0.5

# 3.定义Agent，绑定模型和工具，用函数create_agent
agent = create_agent(
    model="deepseek-v4-flash",
    tools=[get_weather, square_root]
)

# 4.调用Agent，用invoke函数，传入一个Dict，字典要有messages属性，里面是消息的列表
response = agent.invoke({
    "messages": [{"role": "user", "content": "求9的平方根"}]
})
print(response) #
```
